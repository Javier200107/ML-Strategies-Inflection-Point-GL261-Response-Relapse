```python
for mice_group in os.listdir(base_path2):
    mice_group_path = os.path.join(base_path2, mice_group)
    if not os.path.isdir(mice_group_path):
        continue  # Saltar si no es un directorio
    for dayofstudy in os.listdir(mice_group_path):
        dayofstudy_path = os.path.join(mice_group_path, dayofstudy)
        if not os.path.isdir(dayofstudy_path):
            continue  # Saltar si no es un directorio
        for seq_folder in os.listdir(dayofstudy_path):
            if seq_folder.startswith('MRI s'):
                try:
                    seq_path = os.path.join(dayofstudy_path, seq_folder, 'pdata', '1', '2dseq')
                    print(f"Procesando secuencia en: {seq_path}")
                    seq = Dataset(seq_path)  # Cargar el archivo 2dseq
                except Exception as e:
                    print(f"Error al cargar el archivo en {seq_path}: {e}")
                    continue
                # Obtener los datos de la imagen
                image_data = seq._read_data()

                # Verificar la forma de los datos
                #print(f"Forma original de los datos: {image_data.shape}")

                # Asumir que los datos son complejos (real e imaginario), y calcular la magnitud
                image_magnitude = np.sqrt(image_data[..., 0]**2 + image_data[..., 1]**2)

                # Escalar los valores de la imagen a 0-255, que es típico para LUT grays en ImageJ
                image_scaled = 255 * (image_magnitude - np.min(image_magnitude)) / (np.max(image_magnitude) - np.min(image_magnitude))

                # Asegurarse de que sea una matriz 2D adecuada
                image_2d = np.squeeze(image_scaled)

                # Convertir la imagen a un objeto de PIL
                image_2d_uint8 = Image.fromarray(image_2d.astype(np.uint8))

                # Rotar la imagen 90 grados hacia la izquierda
                im_rotated = image_2d_uint8.rotate(-90, expand=True)

                # Si la imagen también está invertida, puedes aplicar un volteo horizontal
                im_corrected = im_rotated.transpose(Image.FLIP_LEFT_RIGHT)

                # Get sequence folder last number
                seq_folder_number = seq_folder.split(' ')[-1]
                # Guardar la imagen en formato JPG con compresión
                output_path = os.path.join(dayofstudy_path, 'MRI images')
                if not os.path.exists(output_path):
                    os.makedirs(output_path)
                # Si existe la carpeta, borrar interior
                """if os.path.exists(output_path):
                    files = os.listdir(output_path)
                    for f in files:
                        os.remove(os.path.join(output_path, f))"""
                #print(f"Guardando imagen en: {output_path}")
                im_corrected.save(output_path +  '/image_{}.jpg'.format(seq_folder_number), 'JPEG', quality=90)  # Ajusta la calidad si es necesario

                print(f"Imagen guardada en formato JPG en: {output_path}\n\n")
```
