```python
import os
import pydicom as dicom
import cv2   
import numpy as np
```

```python
input = "input/External Datasets/Brain tumor progression"
```

```python
for root, dirs, files in os.walk(input):
    for file in files:
        if file.endswith(".dcm"):
            ds = dicom.dcmread(os.path.join(root, file))
            pixel_array_numpy = ds.pixel_array.astype(np.float32)

            # Normalizar la imagen a 8 bits (0-255)
            pixel_array_numpy -= pixel_array_numpy.min()
            pixel_array_numpy /= pixel_array_numpy.max()
            pixel_array_numpy *= 255.0

            # Convertir a uint8
            pixel_array_numpy = pixel_array_numpy.astype(np.uint8)

            # Remove jpeg and png files
            for f in os.listdir(root):
                if f.endswith(".jpeg") or f.endswith(".png"):
                    os.remove(os.path.join(root, f))

            # Guardar la imagen en escala de grises
            new_path = os.path.join(root, file.replace('.dcm', '.jpg'))
            cv2.imwrite(new_path, pixel_array_numpy)

            print(f"Guardado: {new_path}")
```
