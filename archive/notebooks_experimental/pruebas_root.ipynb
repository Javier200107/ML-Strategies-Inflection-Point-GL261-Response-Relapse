```python
import numpy as np
```

```python
a = np.array([1, 2, 3, 4, 5, 6])

a.reshape((2, 3, 1))

print(a)
```

```python

# Creamos un array unidimensional de 3 elementos
array = np.array([5, 10, 15])
print("Original:", array.shape, array)  # (3,) -> Vector 1D

# Aplicamos reshape(1, 3, 1)
reshaped_array = array.reshape(1, 3, 1)
print("Reshaped:", reshaped_array.shape, reshaped_array)
```
